```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('anime-dataset-2023.csv')
data["Episodes"] = data["Episodes"].replace("UNKNOWN", 0).astype(int)
data["Score"] = data["Score"].replace("UNKNOWN", 0).astype(float)
data["Rank"] = data["Rank"].replace("UNKNOWN", 0).astype(int)
data["Scored By"] = data["Scored By"].replace("UNKNOWN", 0).astype(int)
data.to_csv('anime-dataset-2023.csv', index=False)
```
