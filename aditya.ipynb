```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('anime-dataset-2023.csv')
```

```python
#score, episodes, aired, duration , rank, scored by change data types for formatting and analysis
data["Episodes"] = data["Episodes"].replace("UNKNOWN", 0)
data["Episodes"] = data["Episodes"].astype(int)
```

```python
data["Score"] = data["Score"].replace("UNKNOWN", 0)
data["Score"] = data["Score"].astype(float)
```

```python
data["Rank"] = data["Rank"].replace("UNKNOWN", 0)
data["Rank"] = data["Rank"].astype(int)
```

```python
data["Scored By"] = data["Scored By"].replace("UNKNOWN", 0)
data["Scored By"] = data["Scored By"].astype(int)
```
