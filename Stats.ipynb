```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("anime-dataset-2023.csv")
```

This output is exactly what you want to see when doing Exploratory Data Analysis (EDA). By adding that count column, the story your data is telling becomes incredibly clear.

Here are the key analytical insights you can extract from this table, formatted as highly testable concepts for your notes.

1. The "Zero Score" Missing Data Confirmation
Look at the pure Action row at the very top. It has 256 anime, a maximum score of 8.03, and a minimum score of 0.00. The Action, Adventure, Comedy, Fantasy row has 51 anime, a max of 7.65, and a min of 0.00.

Because we see a 0.00 minimum in groups with large sample sizes, it practically guarantees that 0.00 is not a legitimate user rating. It is a placeholder for NaN (Not a Number) or missing data (e.g., an upcoming anime that hasn't aired yet).

2. The "Long Tail" Distribution of Categories
Notice the relationship between the length of the genre name and the count column:

Broad Categories: Simple labels like Action (256 count) or Action, Adventure (57 count) have the highest volume.

Hyper-Specific Categories: The more tags an anime has (e.g., Action, Adventure, Comedy, Drama, Sci-Fi), the lower the count drops, almost exclusively hitting 1, 2, or 3.

In data science, this is known as high cardinality or a "long tail" distribution. Grouping by exact string matches on columns with dozens of comma-separated tags creates hundreds of useless micro-groups.

3. The Singleton Trap Unveiled
Our previous suspicion is completely validated by the count column. Look at the middle section filled with rows that have a count of 1.

For Action, Adventure, Award Winning, Comedy, Mystery, the min is 8.15 and the max is 8.15. Because the count is exactly 1, calculating variance, standard deviation, or mean for this specific group is statistically meaningless.

4. The True Anomalies
Look at Action, Adventure, Comedy, Fantasy, Supernatural. The minimum is 0.00, the maximum is 0.00, and the count is 1. This means there is exactly one show in this highly specific genre combination, and its score data is completely missing.

📚 Data Science Study Notes: Handling Dirty Data
When dealing with dirty datasets containing placeholders like 0.00, you will generally be tested on two primary solutions:

Dropping (Listwise Deletion): Removing any row where the score is 0.00 before running your .groupby(). This is safe if the missing data is a small percentage of the total dataset.

Imputation: Replacing the 0.00 with the median or mean score of that specific genre. This is riskier but preserves your row counts.
