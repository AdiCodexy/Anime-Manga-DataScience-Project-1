```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

data = pd.read_csv("anime-dataset-2023.csv")
df=pd.DataFrame(data)

```

```python
plt.scatter(df['Episodes'], df['Score'], color='red', label='Score vs Episodes')
plt.title('Episodes vs Score')
plt.legend()
plt.xlabel('Episodes')
plt.ylabel('Score')
plt.show()

# This shows that long running animes(500+ episodes) tend to have lower scores. This could be because long running animes tend to have more filler episodes which can lead to a decrease in quality and thus a decrease in score. Additionally, long running animes may also have a larger number of episodes which can lead to a decrease in score due to the sheer volume of content that needs to be produced.
# Whereas short running animes(1-50 episodes) tend to have higher scores. This could be because short running animes tend to have a more focused and concise storyline which can lead to a higher quality and thus a higher score. Additionally, short running animes may also have a smaller number of episodes which can lead to a higher score due to the ability to maintain a consistent level of quality throughout the series.
# Also there are some outliers in the data, such as animes with a score of 10 and 1000+ episodes, which may be due to the fact that these animes are very popular and have a large fanbase, which can lead to a higher score. However, it is important to note that these outliers may not be representative of the overall trend in the data.
```

```python
plt.scatter(df['Rank'], df['Favorites'], color='blue', label='Rank vs Favorites')
plt.title('Rank vs Favorites')
plt.xlabel('Rank')
plt.ylabel('Favorites')
plt.legend()
plt.show()

# This shows that animes with a higher rank tend to have more favorites. This could be because animes with a higher rank are generally considered to be of higher quality and thus tend to have more fans who would add them to their favorites list. Additionally, animes with a higher rank may also have a larger fanbase which can lead to more favorites. However, it is important to note that there may be some outliers in the data, such as animes with a low rank but a high number of favorites, which may be due to the fact that these animes are very popular and have a large fanbase, which can lead to a higher number of favorites.
# Whereas animes with a lower rank tend to have fewer favorites. This could be because animes with a lower rank are generally considered to be of lower quality and thus tend to have fewer fans who would add them to their favorites list. Additionally, animes with a lower rank may also have a smaller fanbase which can lead to fewer favorites. However, it is important to note that there may be some outliers in the data, such as animes with a high rank but a low number of favorites, which may be due to the fact that these animes are not very popular and thus do not have a large fanbase, which can lead to a lower number of favorites.
# But rank is not the only factor that can influence the number of favorites an anime has. Other factors such as the genre, the studio, the source material, and the airing season can also play a role in determining the number of favorites an anime has. Additionally, it is important to note that the number of favorites an anime has may not always be a direct reflection of its quality, as some animes may have a large fanbase but may not necessarily be of high quality.
```

```python
plt.scatter(df['Score'], df['Rank'], color='green', label='Rank vs Score')
plt.title('Rank vs Score')
plt.xlabel('Score')
plt.ylabel('Rank')
plt.legend()
plt.show()

As the score increases, the rank decreases (moves closer to 1).
Anime with scores around 8–9 tend to have the best ranks.

Interpretation: Ranking is heavily determined by score, which is expected since ranks are usually assigned based on ratings.

2. Non-linear relationship
The relationship is not a straight line.
Between scores 5 and 7, rank changes dramatically.
Above 8, even small increases in score lead to large improvements in rank.

Interpretation: Small score differences among highly rated anime can significantly affect rankings.

3. Most top-ranked anime have scores above 7.5
Nearly all anime with ranks close to 1 are concentrated in the score range 7.5–9.5.

Interpretation: To achieve a top rank, an anime generally needs a very high score.

4. Many anime have a score of 0
There's a vertical cluster at Score = 0 spanning many ranks.

Interpretation: These are likely:

Missing ratings
Unrated anime
Data-entry placeholders

For analysis, consider removing or imputing these records:

df = df[df['score'] > 0]
5. Presence of rank = 0 values
There is a horizontal line near Rank = 0.

Interpretation: These are likely missing or unranked entries and may need cleaning:

df = df[df['rank'] > 0]
Key EDA Conclusion

The plot shows a strong negative, non-linear relationship between score and rank. Higher-scoring anime consistently achieve better ranks, with rankings improving rapidly once scores exceed approximately 6–7. The presence of many score and rank values equal to zero suggests missing or unranked records that should be handled during data cleaning.
```
